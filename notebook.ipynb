```python
# Re-run this cell
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
crimes = pd.read_csv("crimes.csv", dtype={"TIME OCC": str})
crimes.head(20)
```

```python
time_str = crimes["TIME OCC"].astype(str).str.zfill(4)
crimes["HOUR OF CRIME"] = pd.to_datetime(time_str, format="%H%M").dt.hour
crimes_sorted_by_hour = crimes.groupby("HOUR OF CRIME").size()


peak_crime_hour = crimes_sorted_by_hour.sort_values(ascending=False)
peak_crime_hour = int(peak_crime_hour.index[0])


night_crimes = crimes[(crimes["HOUR OF CRIME"] >= 22) | (crimes["HOUR OF CRIME"]<4)]
night_crimes_location = night_crimes.groupby("AREA NAME").size()
night_crimes_location = night_crimes_location.sort_values(ascending=False)
peak_night_crime_location = night_crimes_location.head(1)
peak_night_crime_location = str(peak_night_crime_location.index[0])

crimes = crimes[crimes["Vict Age"] >= 0]
age_bins = [0,17,25,34,44,54,64,120]
labels = ["0-17", "18-25", "26-34", "35-44", "45-54", "55-64", "65+"]
crimes["Vict age group"] = pd.cut(crimes["Vict Age"], bins=age_bins, labels=labels, include_lowest=True)
crimes_by_age_group = crimes.groupby("Vict age group").size()
crimes_by_age_group.sort_values(ascending=False,inplace=True)
victim_ages = pd.Series(crimes_by_age_group)
```
